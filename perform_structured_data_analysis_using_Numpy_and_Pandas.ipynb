#Create a 1D NumPy array named temperatures_w1 for Week 1 with the following
values:

```python
!pip install numpy
import numpy as np
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("shape:",temperatures_w1.shape)
print("dtype:",temperatures_w1.dtype)
print("size:",temperatures_w1.size)
print("dimension:",temperatures_w1.ndim)
```

#Array operations

```python
Temp_Fahrenheit=(temperatures_w1 * 9/5) + 32
print(Temp_Fahrenheit)
print("Mean_Temp:",np.mean(Temp_Fahrenheit))
print("Min_Temp:",np.min(Temp_Fahrenheit))
print("Max_Temp:",np.max(Temp_Fahrenheit))
```

Array Slicing and Indexing:

```python
print("1st 3 days temp:",Temp_Fahrenheit[0:3])
print("last 2 days temp:",Temp_Fahrenheit[-2:])
print("Middle 3 days temp:",Temp_Fahrenheit[2:5])
```

2D NUMPY Array

#Pandas

```python
!pip install pandas
import pandas as pd
transactions = pd.DataFrame({"TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
"ProductCategory": ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',

'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
"Region": ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South',
'North'],
"Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})
print(transactions)
print("head:")
print(transactions.head)
print("tail:")
print(transactions.tail)
print("shape:")
print(transactions.shape)
print("Columns:")
print(transactions.columns)
print("dtypes:")
print(transactions.dtypes)
```

```python
print("Productcategory:\n",transactions["ProductCategory"],"Amounts\n",transactions["Amount"])
```

```python
print("Last 3 items:\n",transactions.loc[:,["ProductCategory","Region","Amount"]])
```

```python
print("Last 3 items:\n",transactions.iloc[:,-3:])
```

```python
print(transactions[(transactions["Amount"]>200)&(transactions["Region"]=="North")])
```

```python
print(transactions["ProductCategory"].value_counts())
print("Region_Unique:\n",transactions["Region"].unique())
```

```python
print(transactions.groupby("Region")["Amount"].mean())
```

```python
print(transactions)
transactions.loc[transactions["TransactionID"]==102,"Amount"]=165
```

```python
transactions["Discount"] = transactions["Amount"]*10/100
print(transactions)
```

```python
transactions.drop(transactions[transactions["TransactionID"]==109].index,axis=0,inplace=True)
print(transactions)
```

Pandas Series

```python
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

```python
print("1st rank mark:",marks[0])
print("1st 3 marks:using loc\n",marks.loc[["Rank1","Rank2","Rank3"]])
print("1st 3 marks using iloc:\n",marks.iloc[:2])
mask = marks > 90
print(marks[mask])
```

```python
marks["Rank1"]= 100
marks
marks.drop("Rank5")
CGPA= marks/10
print(CGPA)
```
